# Exercice 1 — Résolution de systèmes linéaires

On résout $Ax = b$ avec $A$ une matrice $4\times4$ **symétrique définie positive (SDP)**.
On compare des méthodes *directes* (LU, Cholesky), qui donnent la solution exacte à l'arrondi
près, et des méthodes *itératives* (Jacobi, Gauss-Seidel), qui approchent la solution étape par étape.

## 1.1 Méthode directe : décomposition LU

### 1.1.1 — Construction d'une matrice SDP $A$ et d'un vecteur $b$

Une matrice $A$ est **symétrique définie positive** si $A = A^\top$ et si $x^\top A x > 0$ pour
tout vecteur non nul, ce qui équivaut à avoir **toutes ses valeurs propres strictement positives**.

Astuce de construction : pour une matrice aléatoire $B$, le produit $B^\top B$ est toujours
symétrique semi-définie positive ; en ajoutant $4 I$ on pousse toutes les valeurs propres
au-dessus de $4$, ce qui garantit le caractère défini positif.

```python
import numpy as np
#Calcul sur les matrices et vecteurs
import scipy.linalg as la
#Algèbre linéaire avancée (LU, Cholesky, QR…)
import time
#Mesurer le temps d'exécution
import matplotlib.pyplot as plt
#Faire des graphiques
import matplotlib.image as mpimg
#Charger des images

np.random.seed(42)
#ça rend les nombres « aléatoires » toujours identiques d'une exécution à l'autre →reproductible.

B = np.random.randn(4, 4)
#Crée une matrice 4 lignes × 4 colonnes remplie de nombres aléatoires (tirés d'une loi normale, c'est-à-dire centrés autour de 0)

A = B.T @ B + 4 * np.eye(4)
#B.T @ B → on multiplie la transposée par B. Cette opération produit toujours une matrice symétrique et semi-définie positive (ses valeurs propres sont ≥ 0).
#+ 4 * np.eye(4) → on ajoute 4 à chaque élément diagonal.Pour pousser toutes les valeurs propres au-dessus de 4 (donc strictement positives).
# Cela garantit que A est Symétrique Définie Positive

b = np.random.randn(4)

# ── Vérifications
print("A =\n", np.round(A, 4))
print("b =", np.round(b, 4))
print("Valeurs propres :", np.round(np.linalg.eigvals(A), 4))
print("Symétrique :", np.allclose(A, A.T))
```

**Interprétation.** Les quatre valeurs propres sont strictement positives
($\approx 13.5,\ 7.7,\ 4.1,\ 4.8$) et `Symétrique = True` : $A$ est bien **SDP**, ce qui légitime
l'usage de Cholesky (section 1.2) en plus de LU.

### 1.1.2 — Décomposition LU avec `scipy.linalg.lu`

`scipy.linalg.lu(A)` renvoie trois matrices :
- **$P$** : matrice de permutation (réordonne les lignes pour le pivotage partiel) ;
- **$L$** : triangulaire inférieure à diagonale unitaire ;
- **$U$** : triangulaire supérieure.

Convention de SciPy : $A = P\,L\,U$, donc $P^\top A = L U$ (car $P^{-1} = P^\top$ pour une permutation).

```python
# scipy.linalg.lu retourne (P, L, U) tels que A = P L U
# (convention scipy : A = PLU, donc PA = L'U' avec réorganisation)
P, L, U = la.lu(A)

print("P =\n", P)
print("L =\n", np.round(L, 4))
print("U =\n", np.round(U, 4))

# Vérification PA = LU : convention scipy → A = P L U
#   donc P^T A = L U  (car P^T = P^{-1} pour les matrices de permutation)
print("Vérif PA == LU :", np.allclose(P.T @ A, L @ U))
```

**Interprétation.** Ici $P$ est la matrice identité : les pivots naturels de $A$ étaient
déjà les plus grands, aucun échange de lignes n'a été nécessaire. La vérification
$P^\top A = L U$ renvoie `True` : la factorisation est correcte.

### 1.1.3 — Résolution par substitutions triangulaires

Une fois $A = PLU$, résoudre $Ax = b$ se ramène à deux systèmes triangulaires (peu coûteux) :

1. permutation du second membre : $\tilde b = P^\top b$ ;
2. **substitution avant** : résoudre $L y = \tilde b$ ;
3. **substitution arrière** : résoudre $U x = y$.

```python
#  A x = b
#  (A = P L U)
#  P L U x = b
# ── Étape 1 : Appliquer la permutation au membre droit
#    P est orthogonale → P^{-1} = P^T
# L U x = P^T b = Pb
Pb = P.T @ b   # vecteur permuté



# on pose y = U x
# ── Étape 2 : Substitution avant — résoudre Ly = Pb
y = la.solve_triangular(L, Pb, lower=True)
#    lower=True indique que L est triangulaire inférieure
# → on trouve y


# ── Étape 3 : Substitution arrière — résoudre Ux = y
x_lu = la.solve_triangular(U, y, lower=False)
#    lower=False (défaut) : triangulaire supérieure

print("la solution du systeme Ax=b =", np.round(x_lu, 6))
```

**Interprétation.** On obtient la solution
$x \approx (-0.1951,\ 0.0478,\ -0.0568,\ -0.1494)$. Elle sera validée juste après.

### 1.1.4 — Vérification avec `numpy.linalg.solve`

`numpy.linalg.solve` s'appuie sur LAPACK (LU avec pivotage). On l'utilise comme référence
pour contrôler notre résolution « manuelle ».

```python
# ── Référence : numpy.linalg.solve utilise LAPACK internalement
#    (LU avec pivotage complet — très robuste)
x_ref = np.linalg.solve(A, b)

# ── Comparaison numérique
print("x_lu  =", np.round(x_lu,  8))
print("x_ref =", np.round(x_ref, 8))
print("Solutions identiques :", np.allclose(x_lu, x_ref))
```

**Interprétation.** `Solutions identiques : True` (test `np.allclose`) : notre résolution
LU par substitutions coïncide avec la routine de référence, à la précision machine près.

## 1.2 Méthode de Cholesky

### 1.2.1 — Résolution avec `cho_factor` / `cho_solve`

Lorsque $A$ est SDP, on peut écrire $A = L L^\top$ (factorisation de Cholesky). Elle est environ
**deux fois moins coûteuse** que LU car elle exploite la symétrie et n'a pas besoin de pivotage.

```python

x_cho = la.cho_solve( la.cho_factor(A), b)

print("x_cho =", np.round(x_cho, 6))
```

**Interprétation.** La solution obtenue est identique à celle de LU :
$x \approx (-0.1951,\ 0.0478,\ -0.0568,\ -0.1494)$.

### 1.2.2 — Comparaison précision / temps

On mesure le temps moyen sur 1000 répétitions et le **résidu** $\lVert Ax - b\rVert_2$
(qualité de la solution) pour chaque méthode.

```python
# ── Benchmark sur 1000 répétitions pour lisser les variations
N = 1000
times_lu, times_cho = [], []

for _ in range(N):
    # --- LU ---
    t0 = time.perf_counter()
    P2, L2, U2 = la.lu(A)
    y2 = la.solve_triangular(L2, P2.T @ b, lower=True)
    x2 = la.solve_triangular(U2, y2)
    times_lu.append(time.perf_counter() - t0)

    # --- Cholesky ---
    t0 = time.perf_counter()
    cf = la.cho_factor(A)
    xc = la.cho_solve(cf, b)
    times_cho.append(time.perf_counter() - t0)

print(f"Temps moyen LU        : {np.mean(times_lu)*1e6:.2f} µs")
print(f"Temps moyen Cholesky  : {np.mean(times_cho)*1e6:.2f} µs")
print(f"Résidu LU             : {np.linalg.norm(A @ x_lu - b):.4e}")
print(f"Résidu Cholesky       : {np.linalg.norm(A @ x_cho - b):.4e}")
```

**Conclusion (1.2).**
- **Précision** : les deux résidus sont de l'ordre de $10^{-16}$, soit la **précision machine**
  ($\varepsilon_{\text{mach}} \approx 2.2\times10^{-16}$). L'écart entre LU et Cholesky est négligeable.
- **Temps** : Cholesky est ici **environ 2 à 3 fois plus rapide** que LU (les valeurs exactes en
  microsecondes dépendent de la machine).
- **Règle pratique** : dès que $A$ est symétrique définie positive, on préfère **Cholesky**.

## 1.3 Méthodes itératives : Jacobi et Gauss-Seidel

### 1.3.1 — Méthode de Jacobi

On décompose $A = D + (L + U)$ où $D$ est la diagonale, $L$ la partie triangulaire inférieure
stricte et $U$ la partie triangulaire supérieure stricte. Jacobi met à jour **toutes les
composantes simultanément** à partir de l'itération précédente :

$$x^{(k+1)} = D^{-1}\big(b - (L+U)\,x^{(k)}\big).$$

Le critère d'arrêt porte sur la norme de la correction $\lVert x^{(k+1)} - x^{(k)}\rVert_2 < \text{tol}$.

```python
def jacobi(A, b, tol=1e-6, max_iter=100):
    """
    Résolution de Ax = b par la méthode de Jacobi.

    Paramètres
    ----------
    A        : ndarray (n,n)  — matrice à diagonale dominante ou SPD
    b        : ndarray (n,)   — vecteur second membre
    tol      : float          — tolérance de convergence (défaut 1e-6)
    max_iter : int            — nombre max d'itérations (défaut 100)

    Retourne
    --------
    x     : solution approchée
    k+1   : nombre d'itérations effectuées
    hist  : liste des normes de correction à chaque itération
    """
    n    = len(b)
    x    = np.zeros(n)          # initialisation x^(0) = 0
    D    = np.diag(A)           # extraction de la diagonale
    R    = A - np.diag(D)       # R = L + U (hors-diagonale)
    hist = []

    for k in range(max_iter):
        # Mise à jour simultanée de toutes les composantes
        #  x_new_i = (b_i - Σ_{j≠i} A_ij x_j^(k)) / A_ii
        #  En vectoriel : x_new = (b - R @ x) / D
        x_new = (b - R @ x) / D

        # Norme de la correction (critère d'arrêt)
        correction = np.linalg.norm(x_new - x)
        hist.append(correction)

        if correction < tol:
            return x_new, k + 1, hist    # convergé
        x = x_new

    return x, max_iter, hist              # non convergé (max atteint)
```

### 1.3.2 — Méthode de Gauss-Seidel

Gauss-Seidel met à jour les composantes **séquentiellement** : pour calculer $x_i^{(k+1)}$,
on réutilise immédiatement les valeurs $x_1^{(k+1)}, \dots, x_{i-1}^{(k+1)}$ déjà recalculées.
Cette « information fraîche » accélère en général la convergence (souvent d'un facteur ~2).

```python
def gauss_seidel(A, b, tol=1e-6, max_iter=100):
    """
    Résolution de Ax = b par la méthode de Gauss-Seidel.

    Paramètres et retour identiques à jacobi().
    Clé : utilisation immédiate des nouvelles valeurs dans la même itération.
    """
    n    = len(b)
    x    = np.zeros(n)    # initialisation x^(0) = 0
    hist = []

    for k in range(max_iter):
        x_old = x.copy()  # sauvegarder x^(k) pour le critère d'arrêt

        for i in range(n):
            # somme de TOUS les termes hors-diagonale
            #   j < i → x[j] est déjà x^(k+1)  (valeur fraîche)
            #   j > i → x[j] est encore x^(k)  (ancienne valeur)
            s = sum(A[i, j] * x[j] for j in range(n) if j != i)
            x[i] = (b[i] - s) / A[i, i]
            # x[i] immédiatement disponible pour les prochains j > i

        correction = np.linalg.norm(x - x_old)
        hist.append(correction)

        if correction < tol:
            return x, k + 1, hist

    return x, max_iter, hist
```

### 1.3.3 — Test des deux méthodes (tolérance $10^{-6}$, max 100 itérations)

On lance les deux solveurs sur le même système et on compare à la solution de référence (LU).

```python
# ── Exécution des deux méthodes
x_jac, iter_jac, hist_jac = jacobi(A, b, tol=1e-6, max_iter=100)
x_gs,  iter_gs,  hist_gs  = gauss_seidel(A, b, tol=1e-6, max_iter=100)

# ── Résultats
print(f"=== Jacobi ===")
print(f"  Itérations : {iter_jac}")
print(f"  Solution   : {np.round(x_jac, 6)}")
print(f"  Résidu     : {np.linalg.norm(A @ x_jac - b):.2e}")

print(f"\n=== Gauss-Seidel ===")
print(f"  Itérations : {iter_gs}")
print(f"  Solution   : {np.round(x_gs, 6)}")
print(f"  Résidu     : {np.linalg.norm(A @ x_gs - b):.2e}")

# ── Référence LU pour valider
print(f"\n=== Référence LU ===")
print(f"  Solution   : {np.round(x_lu, 6)}")
```

**Interprétation.** Les deux méthodes convergent vers la **même** solution que LU.
Jacobi a besoin de **19 itérations**, Gauss-Seidel de seulement **10**.

### 1.3.4 — Comparaison du nombre d'itérations et de la précision

Le tableau et le graphe (échelle logarithmique) montrent la décroissance de la correction
à chaque itération.

```python
# ── Tableau de convergence
print(f"{'Itér':>5} | {'‖correction‖ Jacobi':>20} | {'‖correction‖ GS':>16}")
print("-"*48)
for k in range(max(len(hist_jac), len(hist_gs))):
    j_val = f"{hist_jac[k]:.6f}" if k < len(hist_jac) else "convergé  "
    g_val = f"{hist_gs[k]:.6f}"  if k < len(hist_gs)  else "convergé  "
    print(f"{k+1:>5} | {j_val:>20} | {g_val:>16}")

# ── Visualisation de la convergence
plt.figure(figsize=(8, 4))
plt.semilogy(range(1, iter_jac+1), hist_jac, 'o-', label='Jacobi', color='#ff9f43')
plt.semilogy(range(1, iter_gs+1),  hist_gs,  's-', label='Gauss-Seidel', color='#38d9a9')
plt.axhline(1e-6, linestyle='--', color='gray', label='Tolérance 10⁻⁶')
plt.xlabel('Itération')
plt.ylabel('‖correction‖₂ (log)')
plt.legend()
plt.title('Convergence Jacobi vs Gauss-Seidel')
plt.tight_layout()
plt.savefig('convergence.png', dpi=150)
plt.show()
```

**Analyse comparative (1.3).**
- **Itérations** : Gauss-Seidel converge en **10** itérations contre **19** pour Jacobi (~2× moins).
- **Précision** : à convergence, le résidu de Gauss-Seidel ($\approx 1.6\times10^{-7}$) est plus
  faible que celui de Jacobi ($\approx 3.9\times10^{-6}$).
- **Convergence garantie** : ici $A$ est SDP (et à diagonale dominante), ce qui assure la
  convergence des deux méthodes. Sur la courbe, la pente de Gauss-Seidel est plus raide.

# Exercice 2 — Décompositions SVD et QR

## 2.1 Décomposition en valeurs singulières (SVD)

### 2.1.1 — Matrice de données $D$ (10 × 5)

On simule un jeu de données : **10 étudiants** notés sur **5 matières** (notes entières entre 0 et 19).

```python

np.random.seed(0)


D = np.random.randint(0, 20, size=(10, 5)).astype(float)

print("D (10 étudiants × 5 matières) :")
print(D)
```

**Interprétation.** Chaque **ligne** est un étudiant, chaque **colonne** une matière.
Cette matrice servira aussi à l'ACP (exercice 3).

### 2.1.2 — Calcul de la SVD

`numpy.linalg.svd` factorise $D = U\,\Sigma\,V^\top$. On utilise `full_matrices=False`
(SVD « économique ») : $U$ est $10\times5$, les valeurs singulières $\sigma_i$ forment un vecteur
de taille 5 (décroissant), et $V^\top$ est $5\times5$.

```python
# ── SVD économique (thin SVD) : full_matrices=False

U_svd, S_svd, Vt_svd = np.linalg.svd(D, full_matrices=False)

print("U :")
print(U_svd)
print("S :")
print(S_svd)
print("Vt :")
print(Vt_svd)

print("Formes : U=", U_svd.shape, " S=", S_svd.shape, " Vt=", Vt_svd.shape)
```

**Interprétation.** Les valeurs singulières $\Sigma \approx (74.7,\ 22.6,\ 14.7,\ 8.7,\ 6.3)$
décroissent rapidement : la première concentre à elle seule **~87 %** de l'énergie
($\sigma_1^2 / \sum_i \sigma_i^2$). L'information est donc très concentrée sur les premières composantes.

### 2.1.3 — Reconstruction et vérification

On reconstruit $D$ à partir de toutes ses composantes : $D_{\text{rec}} = U\,\mathrm{diag}(\Sigma)\,V^\top$,
puis on mesure l'erreur de Frobenius $\lVert D - D_{\text{rec}}\rVert_F$.

```python
# ── Reconstruction complète : D_rec = U @ diag(S) @ Vt
D_rec = U_svd @ np.diag(S_svd) @ Vt_svd
print("D_rec (reconstruction complète) :")
print(D_rec)

err= np.linalg.norm(D - D_rec, 'fro')

print(f"‖D - D_rec‖ = {err:.4e}")

print("Reconstruction exacte     :", np.allclose(D, D_rec))
```

**Interprétation.** L'erreur est $\approx 5.8\times10^{-14}$ (et `Reconstruction exacte = True`) :
en gardant les 5 composantes, la SVD reconstruit $D$ de manière **numériquement exacte**.

### 2.1.4 — Approximation de rang 2

On ne conserve que les **2 plus grandes** valeurs singulières :
$D_2 = U_{:,:2}\,\mathrm{diag}(\sigma_{1:2})\,V^\top_{:2,:}$. C'est la meilleure approximation de rang 2
au sens de Frobenius (théorème d'**Eckart-Young**).

```python
# ── Approximation de rang 2 : ne garder que les 2 premières composantes
k = 2
D_approx = U_svd[:, :k] @ np.diag(S_svd[:k]) @ Vt_svd[:k, :]
print(D_approx )
# ── Erreur d'approximation
err_approx = np.linalg.norm(D - D_approx, 'fro')
print(f"Erreur relative  = {err_approx / np.linalg.norm(D,'fro') * 100:.2f}%")
```

**Interprétation.** L'erreur relative de Frobenius est d'environ **22.7 %**. Autrement dit,
avec seulement **2 composantes sur 5** on conserve $\approx 95\%$ de l'énergie
($1 - 0.227^2 \approx 0.95$) : un excellent compromis taille / fidélité, qui justifie la
réduction de dimension utilisée dans l'ACP.

## 2.2 Décomposition QR

### 2.2.1 — Décomposition QR de $D$

`scipy.linalg.qr(D)` factorise $D = Q R$ avec $Q$ orthogonale ($10\times10$) et $R$ triangulaire
supérieure ($10\times5$). Les pivots (diagonale de $R$) renseignent sur le conditionnement.

```python
# ── Décomposition QR de D (10×5)
#    Q : matrice orthogonale (10×10) — base orthonormale de ℝ¹⁰
#    R : matrice triangulaire supérieure (10×5)
Q, R = la.qr(D)

print("Formes : Q =", Q.shape, "  R =", R.shape)
print("\nDiagonale de R (pivots QR) :", np.round(np.diag(R[:5,:]), 4))
print("\nR (5×5, coin supérieur gauche) :\n", np.round(R[:5,:5], 4))
```

### 2.2.2 — Vérification de l'orthogonalité et de la factorisation

On vérifie que $Q^\top Q = I$ (colonnes orthonormées) et que $Q R = D$.

```python
# ── Vérification 1 : Q est-elle orthogonale ?
#    Q orthogonale ⟺ Q^T Q = I_n (produit de colonnes orthonormées)
I_approx = Q.T @ Q
print("QᵀQ ≈ I ?", np.allclose(I_approx, np.eye(Q.shape[1])))

# ── Vérification 2 : D = QR ?
print("QR = D ?", np.allclose(Q @ R, D))
```

**Interprétation.** Les deux tests renvoient `True` : $Q$ est bien **orthogonale** et le produit
$QR$ restitue exactement $D$. La QR est l'outil de choix pour les **moindres carrés** et la
**stabilité numérique** (orthogonalisation).

# Exercice 3 — Application à l'analyse de données

## 3.1 Analyse en Composantes Principales (ACP)

On applique l'ACP à la matrice $D$ de l'exercice 2 (10 étudiants × 5 matières). L'ACP via SVD
consiste à : (1) centrer les données, (2) faire la SVD de la matrice centrée, (3) projeter sur
les premières composantes principales.

### 3.1.1 — Centrage des données

On soustrait la moyenne de chaque colonne (matière). Le centrage est indispensable : l'ACP
analyse la **variance autour de la moyenne**, pas les valeurs brutes.

```python
# ── Calcul de la moyenne de chaque colonne (matière)
D_mean = D.mean(axis=0)   # shape (5,) — une moyenne par matière

# ── Centrage : broadcasting soustrait D_mean de chaque ligne
#    D_centered[i, j] = D[i, j] - D_mean[j]
D_centered = D - D_mean

print("Moyenne par matière (avant centrage) :", np.round(D_mean, 3))
print("Moyenne après centrage               :", np.round(D_centered.mean(axis=0), 10))
print("\nD_centered (5 premières lignes) :\n", np.round(D_centered[:5], 3))
```

**Interprétation.** Les moyennes par matière étaient $\approx (9.7,\ 10.9,\ 7.6,\ 12.8,\ 5.2)$.
Après centrage, la moyenne de chaque colonne vaut $0$ (à l'arrondi près) : les données sont prêtes.

### 3.1.2 — SVD sur les données centrées

Les **lignes** de $V^\top$ sont les **composantes principales** (axes), et les valeurs singulières
au carré sont proportionnelles à la **variance** portée par chaque axe.

```python
# ── SVD sur la matrice centrée
#    Les vecteurs lignes de Vt_acp sont les composantes principales
U_acp, S_acp, Vt_acp = np.linalg.svd(D_centered, full_matrices=False)
print ("U_acp")
print (U_acp)
print ("S_acp")
print (S_acp)
print ("Vt_acp")
print (Vt_acp)
```

**Interprétation.** Les valeurs singulières centrées sont $\approx (32.1,\ 22.2,\ 13.7,\ 8.2,\ 6.0)$.
La variance expliquée vaut donc **~56.7 %** pour CP1, et **~84.0 %** pour CP1+CP2 cumulées :
deux axes suffisent à résumer l'essentiel de l'information (voir l'éboulis ci-dessous).

### 3.1.3 — Projection sur les 2 premières composantes principales

On projette chaque étudiant sur le plan (CP1, CP2) : $Z = D_{\text{centré}}\,V_{:,:2}$.

```python
# ── Projection : Z = D_centered @ V[:, :2]
#    V = Vt_acp.T  →  les 2 premières CP sont les 2 premières lignes de Vt_acp
Z = D_centered @ Vt_acp[:2].T   # shape : (10, 2)

print("Coordonnées des 10 étudiants dans le plan CP1/CP2 :")
print(f"{'Étudiant':>9} | {'CP1':>0} | {'CP2':>0}")
print("-"*32)
for i, (z1, z2) in enumerate(Z):
    print(f"    Ét.{i+1:>2}  | {z1:0.3f} | {z2:0.3f}")
```

**Interprétation.** Chaque étudiant est résumé par 2 coordonnées au lieu de 5 notes.
Ces coordonnées vont nous permettre de visualiser et d'interpréter les profils.

### 3.1.4 — Visualisation

À gauche : nuage des étudiants dans le plan CP1/CP2. À droite : **éboulis** (variance expliquée
par composante et variance cumulée).

```python
# Variance expliquee par chaque composante principale (proportionnelle a sigma^2)
var_ratio = S_acp**2 / np.sum(S_acp**2) * 100
cum_ratio = np.cumsum(var_ratio)

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# --- Graphe 1 : projection des etudiants dans le plan CP1/CP2 ---
ax1 = axes[0]
colors = plt.cm.tab10(np.linspace(0, 1, 10))
for i, (z1, z2) in enumerate(Z):
    ax1.scatter(z1, z2, s=110, color=colors[i], zorder=3)
    ax1.annotate(f'Et.{i+1}', (z1, z2), xytext=(5, 5),
                 textcoords='offset points', fontsize=9, color=colors[i])
ax1.axhline(0, color='gray', lw=0.8, ls='--')
ax1.axvline(0, color='gray', lw=0.8, ls='--')
ax1.set_xlabel(f'CP1  ({var_ratio[0]:.1f} % de variance)')
ax1.set_ylabel(f'CP2  ({var_ratio[1]:.1f} % de variance)')
ax1.set_title('Projection ACP - plan CP1 / CP2')
ax1.grid(True, alpha=0.3)

# --- Graphe 2 : eboulis (scree plot) de la variance expliquee ---
ax2 = axes[1]
comp = np.arange(1, len(var_ratio) + 1)
ax2.bar(comp, var_ratio, color='#4dabf7', label='Variance par composante')
ax2.plot(comp, cum_ratio, 'o-', color='#e8590c', label='Variance cumulee')
for x, y in zip(comp, cum_ratio):
    ax2.annotate(f'{y:.0f}%', (x, y), textcoords='offset points',
                 xytext=(0, 8), ha='center', fontsize=8, color='#e8590c')
ax2.set_xlabel('Composante principale')
ax2.set_ylabel('% de variance')
ax2.set_title("Eboulis des valeurs singulieres (variance expliquee)")
ax2.set_xticks(comp)
ax2.set_ylim(0, 105)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('acp_visualisation.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Interprétation des composantes.**
- **CP1 (~56.7 %)** = **niveau général**. Tous les coefficients de l'axe sont de même signe :
  l'axe oppose les étudiants forts aux étudiants faibles. Par exemple l'étudiant 7
  ($[17,19,19,19,14]$, le meilleur) et l'étudiant 10 ($[0,0,4,5,6]$, le plus faible) se retrouvent
  aux deux extrêmes de CP1.
- **CP2 (~27.3 %)** = **contraste entre matières** (matières 1-2 contre matières 3-4) :
  il distingue les profils « forts dans certaines matières / faibles dans d'autres » à niveau global
  comparable (ex. l'étudiant 2 $[7,9,19,18,4]$ s'oppose à l'étudiant 1 $[12,15,0,3,3]$).
- L'éboulis confirme que **2 composantes** capturent ~84 % de la variance : la projection 2D est fidèle.

## 3.2 Compression d'image par SVD

Une image en niveaux de gris est une matrice de pixels. En ne gardant que les $k$ premières
valeurs singulières, on obtient une version compressée : on ne stocke que $k\,(m+n+1)$ nombres
au lieu de $m \times n$.

### 3.2.1 — Chargement et conversion en niveaux de gris

On charge une image (un fichier local `image.png` s'il est présent, sinon une photo d'exemple de
matplotlib) et on la convertit en niveaux de gris.

```python
import matplotlib.image as mpimg
import matplotlib.cbook as cbook

# --- Chargement de l'image (cf. enonce : matplotlib.image) ---
# 1) on essaie un fichier local 'image.png' ;
# 2) a defaut, on prend une photo d'exemple fournie par matplotlib (600x512),
#    afin que le notebook reste executable partout, sans telechargement.
try:
    img_rgb = mpimg.imread('image.png')
    source = "image.png (fichier local)"
except FileNotFoundError:
    with cbook.get_sample_data('grace_hopper.jpg') as f:
        img_rgb = mpimg.imread(f)
    source = "image d'exemple matplotlib (grace_hopper.jpg, 600x512)"

# Normalisation des intensites dans [0, 1] (les images 8 bits sont dans [0, 255])
img_rgb = img_rgb.astype(float)
if img_rgb.max() > 1.0:
    img_rgb = img_rgb / 255.0

# --- Conversion en niveaux de gris (formule perceptuelle ITU-R BT.601) ---
# Ponderation calquee sur la sensibilite de l'oeil : vert (59%) > rouge (30%) > bleu (11%)
if img_rgb.ndim == 2:                 # image deja en niveaux de gris
    img_gray = img_rgb
else:                                 # image couleur RGB(A)
    img_gray = (0.2989 * img_rgb[:, :, 0]
              + 0.5870 * img_rgb[:, :, 1]
              + 0.1140 * img_rgb[:, :, 2])

print(f"Source                : {source}")
print(f"Dimensions image RGB  : {img_rgb.shape}")
print(f"Dimensions image gris : {img_gray.shape}")
print(f"Plage de valeurs      : [{img_gray.min():.3f}, {img_gray.max():.3f}]")
```

**Interprétation.** L'image est une matrice $600\times512$ de valeurs dans $[0, 1]$ (gris).

### 3.2.2 — SVD de l'image

```python
# ── SVD de la matrice image (m×n pixels)
#    full_matrices=False → SVD économique, évite de stocker U entière
U_img, S_img, Vt_img = np.linalg.svd(img_gray, full_matrices=False)

m, n = img_gray.shape
print(f"Forme U  : {U_img.shape}")
print(f"Forme S  : {S_img.shape}")
print(f"Forme Vt : {Vt_img.shape}")
print(f"\n10 premières valeurs singulières : {np.round(S_img[:10], 2)}")
print(f"Énergie capturée par σ₁          : {S_img[0]**2 / (S_img**2).sum() * 100:.1f}%")
```

**Interprétation.** Le rang de l'image est 512. La première valeur singulière capture déjà
**~73 %** de l'énergie : l'image est largement compressible.

### 3.2.3 — Reconstruction à différents rangs

On teste les rangs 10, 50, 100 et le rang complet. Pour chaque rang on affiche le **taux de
stockage** et le **PSNR** (Peak Signal-to-Noise Ratio, en dB — plus c'est élevé, mieux c'est).

```python
def svd_compress(U, S, Vt, k):
    """Reconstruit l'image avec les k premières valeurs singulières."""
    img_k = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
    # Clamp entre 0 et 1 (artefacts possibles hors plage)
    return np.clip(img_k, 0, 1)

# ── Calcul du ratio de compression pour chaque rang
def compression_ratio(m, n, k):
    """Proportion d'éléments stockés par rapport à l'image originale."""
    return k * (m + n + 1) / (m * n)

ranks = [10, 50, 100, len(S_img)]   # 4 rangs à tester
for k in ranks:
    ratio = compression_ratio(m, n, k)
    img_k = svd_compress(U_img, S_img, Vt_img, k)
    err   = np.linalg.norm(img_gray - img_k, 'fro')
    psnr  = 20 * np.log10(1.0 / np.sqrt(np.mean((img_gray-img_k)**2)))
    print(f"rang={k:4d} | stockage={ratio*100:5.1f}% | ‖err‖_F={err:7.3f} | PSNR={psnr:6.2f} dB")
```

### 3.2.4 — Visualisation

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, k in zip(axes, ranks):
    img_k = svd_compress(U_img, S_img, Vt_img, k)
    ratio = compression_ratio(m, n, k)
    psnr  = 20 * np.log10(1.0 / np.sqrt(np.mean((img_gray-img_k)**2)))

    ax.imshow(img_k, cmap='gray', vmin=0, vmax=1)
    ax.axis('off')

    if k == len(S_img):
        title = "Originale\n(rang 512, 100%)"
    else:
        title = f"rang = {k}\n({ratio*100:.0f}% stocké — PSNR {psnr:.1f} dB)"
    ax.set_title(title, fontsize=9)

plt.suptitle('Compression SVD : impact du rang sur la qualité', fontsize=11)
plt.tight_layout()
plt.savefig('compression_svd.png', dpi=150, bbox_inches='tight')
plt.show()
```

**Interprétation (compression).**
- **rang 10** (~3.6 % de stockage) : seules les grandes structures subsistent, image floue,
  PSNR ~19.5 dB. Utilisable pour une vignette.
- **rang 50** (~18.1 %) : image reconnaissable, PSNR ~27.2 dB — seuil acceptable.
- **rang 100** (~36.2 %) : bonne qualité visuelle, PSNR ~32 dB, soit une compression ~3×.
- **règle pratique** : un PSNR > 30 dB est généralement jugé « bonne qualité ». Garder le rang complet
  (512) coûterait *plus* que l'image d'origine (~185 % de stockage) : tout l'intérêt est de tronquer.

# Questions de synthèse

## 1. Performances (temps, précision) : méthodes directes vs itératives

**Précision.** Les méthodes directes (LU, Cholesky) atteignent la **précision machine**
(résidu $\approx 2\text{-}4\times10^{-16} \approx \varepsilon_{\text{mach}}$) : elles résolvent le
système de façon exacte à l'arrondi flottant près. Les méthodes itératives s'arrêtent dès que la
correction passe sous la tolérance $\varepsilon = 10^{-6}$, d'où un résidu plus élevé
($\approx 3.9\times10^{-6}$ pour Jacobi, $\approx 1.6\times10^{-7}$ pour Gauss-Seidel).

**Temps.** Sur notre matrice $4\times4$ dense, Cholesky est ~2-3× plus rapide que LU, et les
méthodes itératives effectuent 10 à 19 itérations en $O(n^2)$ chacune. Mais pour de **grandes
matrices creuses** ($n \sim 10^4\text{-}10^6$), les méthodes itératives deviennent indispensables :
chaque itération ne coûte que $O(\text{nnz})$ (nombre d'éléments non nuls), alors que la
factorisation LU dense coûte $O(n^3)$ en temps et peut demander $O(n^2)$ en mémoire.

**Conclusion.** Petites/moyennes matrices denses → **LU ou Cholesky** (précision machine garantie).
Grandes matrices creuses ou systèmes issus d'EDP → **Jacobi, Gauss-Seidel**, ou des méthodes plus
performantes (gradient conjugué, GMRES).

## 2. Impact de la réduction de dimension via SVD

**Optimalité (Eckart-Young).** Parmi toutes les matrices de rang $\le k$, l'approximation
$D_k = U_k \Sigma_k V_k^\top$ **minimise** l'erreur (normes de Frobenius et spectrale). La SVD donne
donc la *meilleure* compression possible à rang fixé.

**Effet concret sur $D$.** $\sigma_1 \approx 74.7$ capte à elle seule ~87 % de l'énergie : les notes
des étudiants sont fortement corrélées entre matières (un « niveau général » explique l'essentiel).
Garder **2 composantes** conserve ~95 % de l'énergie, ce qui suffit pour la visualisation et
l'analyse exploratoire.

**Avantages.** Réduction du bruit (les petites $\sigma$ capturent souvent du bruit), suppression de
la multicolinéarité, lutte contre la « malédiction de la dimension », et mise en évidence de
structures latentes.

**Limite.** L'erreur de rang 2 (~22.7 % en norme de Frobenius relative) signifie que certaines
nuances individuelles sont perdues. Le choix du rang est un **compromis fidélité / complexité**.

## 3. Avantages et limites de l'ACP

**Avantages.** L'ACP est **non supervisée**, sans hyperparamètre critique, et produit des
composantes **orthogonales** (décorrélées) qui éliminent la multicolinéarité. Elle maximise la
variance conservée composante par composante (propriété d'optimalité de la SVD), réduit le coût des
algorithmes en aval (k-means, SVM, régression) et offre une **visualisation 2D/3D** naturelle. Sur
$D$, les 2 premières composantes capturent ~84 % de la variance : une synthèse exploitable.

**Limites.** L'ACP est **linéaire** : elle ne capte pas les structures non linéaires (spirales,
clusters concentriques). Elle est **sensible à l'échelle** : sans standardisation, les matières à
grande variance dominent. Les composantes ne sont pas toujours **interprétables** (CP1 est une
combinaison de toutes les matières, pas une matière précise). Enfin l'ACP suppose que *variance =
information utile*, ce qui est faux pour certaines tâches (les directions de faible variance peuvent
être discriminantes).

**Alternatives.** Pour dépasser ces limites : **t-SNE / UMAP** (structures non linéaires, mais sans
reconstruction), **Kernel PCA** (ACP dans un espace de caractéristiques via un noyau), **ICA**
(sources indépendantes plutôt que simplement décorrélées).
