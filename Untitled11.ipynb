1.1.1
Une matrice A est symétrique définie positive si et seulement si xᵀAx > 0 pour tout vecteur non nul, ce qui équivaut à avoir toutes ses valeurs propres strictement positives.

```python
import numpy as np
#Calcul sur les matrices et vecteurs
import scipy.linalg as la
#Algèbre linéaire avancée (LU, Cholesky, QR…)
import time
#Mesurer le temps d'exécution
import matplotlib.pyplot as plt
#Faire des graphiques
import matplotlib.image as mpimg
#Charger des images

np.random.seed(42)
#ça rend les nombres « aléatoires » toujours identiques d'une exécution à l'autre →reproductible.

B = np.random.randn(4, 4)
#Crée une matrice 4 lignes × 4 colonnes remplie de nombres aléatoires (tirés d'une loi normale, c'est-à-dire centrés autour de 0)

A = B.T @ B + 4 * np.eye(4)
#B.T @ B → on multiplie la transposée par B. Cette opération produit toujours une matrice symétrique et semi-définie positive (ses valeurs propres sont ≥ 0).
#+ 4 * np.eye(4) → on ajoute 4 à chaque élément diagonal.Pour pousser toutes les valeurs propres au-dessus de 4 (donc strictement positives).
# Cela garantit que A est Symétrique Définie Positive

b = np.random.randn(4)

# ── Vérifications
print("A =\n", np.round(A, 4))
print("b =", np.round(b, 4))
print("Valeurs propres :", np.round(np.linalg.eigvals(A), 4))
print("Symétrique :", np.allclose(A, A.T))
```

1.1.2 La fonction scipy.linalg.lu(A) retourne trois matrices : P (permutation, réarrange les lignes pour le pivotage partiel), L (triangulaire inférieure, diagonale de 1), U (triangulaire supérieure). La relation vérifiée est PA = LU, équivalente à A = P⁻¹LU = PᵀLU.

```python
# scipy.linalg.lu retourne (P, L, U) tels que A = P L U
# (convention scipy : A = PLU, donc PA = L'U' avec réorganisation)
P, L, U = la.lu(A)

print("P =\n", P)
print("L =\n", np.round(L, 4))
print("U =\n", np.round(U, 4))

# Vérification PA = LU : convention scipy → A = P L U
#   donc P^T A = L U  (car P^T = P^{-1} pour les matrices de permutation)
print("Vérif PA == LU :", np.allclose(P.T @ A, L @ U))
```

1.1.3 La résolution se décompose en deux étapes :

(1) substitution avant — résoudre Ly = Pᵀb

(2) substitution arrière — résoudre Ux = y

```python
#  A x = b
#  (A = P L U)
#  P L U x = b
# ── Étape 1 : Appliquer la permutation au membre droit
#    P est orthogonale → P^{-1} = P^T
# L U x = P^T b = Pb
Pb = P.T @ b   # vecteur permuté



# on pose y = U x
# ── Étape 2 : Substitution avant — résoudre Ly = Pb
y = la.solve_triangular(L, Pb, lower=True)
#    lower=True indique que L est triangulaire inférieure
# → on trouve y


# ── Étape 3 : Substitution arrière — résoudre Ux = y
x_lu = la.solve_triangular(U, y, lower=False)
#    lower=False (défaut) : triangulaire supérieure

print("la solution du systeme Ax=b =", np.round(x_lu, 6))
```

1.3.1 Dans ce contexte, D = diagonale de A, L = partie triangulaire inférieure stricte, U = partie triangulaire supérieure stricte. Jacobi met à jour toutes les composantes simultanément avec les valeurs de l'itération précédente : x⁽ᵏ⁾.

```python
def jacobi(A, b, tol=1e-6, max_iter=100):
    """
    Résolution de Ax = b par la méthode de Jacobi.

    Paramètres
    ----------
    A        : ndarray (n,n)  — matrice à diagonale dominante ou SPD
    b        : ndarray (n,)   — vecteur second membre
    tol      : float          — tolérance de convergence (défaut 1e-6)
    max_iter : int            — nombre max d'itérations (défaut 100)

    Retourne
    --------
    x     : solution approchée
    k+1   : nombre d'itérations effectuées
    hist  : liste des normes de correction à chaque itération
    """
    n    = len(b)
    x    = np.zeros(n)          # initialisation x^(0) = 0
    D    = np.diag(A)           # extraction de la diagonale
    R    = A - np.diag(D)       # R = L + U (hors-diagonale)
    hist = []

    for k in range(max_iter):
        # Mise à jour simultanée de toutes les composantes
        #  x_new_i = (b_i - Σ_{j≠i} A_ij x_j^(k)) / A_ii
        #  En vectoriel : x_new = (b - R @ x) / D
        x_new = (b - R @ x) / D

        # Norme de la correction (critère d'arrêt)
        correction = np.linalg.norm(x_new - x)
        hist.append(correction)

        if correction < tol:
            return x_new, k + 1, hist    # convergé
        x = x_new

    return x, max_iter, hist              # non convergé (max atteint)
```

1.3.2 Gauss-Seidel met à jour les composantes séquentiellement : pour calculer x_i⁽ᵏ⁺¹⁾, on utilise déjà les valeurs x_1⁽ᵏ⁺¹⁾, ..., x_{i-1}⁽ᵏ⁺¹⁾ (fraîchement calculées) plutôt que les anciennes. Cela accélère typiquement la convergence d'un facteur 2.

```python
def gauss_seidel(A, b, tol=1e-6, max_iter=100):
    """
    Résolution de Ax = b par la méthode de Gauss-Seidel.

    Paramètres et retour identiques à jacobi().
    Clé : utilisation immédiate des nouvelles valeurs dans la même itération.
    """
    n    = len(b)
    x    = np.zeros(n)    # initialisation x^(0) = 0
    hist = []

    for k in range(max_iter):
        x_old = x.copy()  # sauvegarder x^(k) pour le critère d'arrêt

        for i in range(n):
            # somme de TOUS les termes hors-diagonale
            #   j < i → x[j] est déjà x^(k+1)  (valeur fraîche)
            #   j > i → x[j] est encore x^(k)  (ancienne valeur)
            s = sum(A[i, j] * x[j] for j in range(n) if j != i)
            x[i] = (b[i] - s) / A[i, i]
            # x[i] immédiatement disponible pour les prochains j > i

        correction = np.linalg.norm(x - x_old)
        hist.append(correction)

        if correction < tol:
            return x, k + 1, hist

    return x, max_iter, hist
```

```python
# ── Exécution des deux méthodes
x_jac, iter_jac, hist_jac = jacobi(A, b, tol=1e-6, max_iter=100)
x_gs,  iter_gs,  hist_gs  = gauss_seidel(A, b, tol=1e-6, max_iter=100)

# ── Résultats
print(f"=== Jacobi ===")
print(f"  Itérations : {iter_jac}")
print(f"  Solution   : {np.round(x_jac, 6)}")
print(f"  Résidu     : {np.linalg.norm(A @ x_jac - b):.2e}")

print(f"\n=== Gauss-Seidel ===")
print(f"  Itérations : {iter_gs}")
print(f"  Solution   : {np.round(x_gs, 6)}")
print(f"  Résidu     : {np.linalg.norm(A @ x_gs - b):.2e}")

# ── Référence LU pour valider
print(f"\n=== Référence LU ===")
print(f"  Solution   : {np.round(x_lu, 6)}")
```

```python
# ── SVD de la matrice image (m×n pixels)
#    full_matrices=False → SVD économique, évite de stocker U entière
U_img, S_img, Vt_img = np.linalg.svd(img_gray, full_matrices=False)

m, n = img_gray.shape
print(f"Forme U  : {U_img.shape}")
print(f"Forme S  : {S_img.shape}")
print(f"Forme Vt : {Vt_img.shape}")
print(f"\n10 premières valeurs singulières : {np.round(S_img[:10], 2)}")
print(f"Énergie capturée par σ₁          : {S_img[0]**2 / (S_img**2).sum() * 100:.1f}%")
```

```python
def svd_compress(U, S, Vt, k):
    """Reconstruit l'image avec les k premières valeurs singulières."""
    img_k = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
    # Clamp entre 0 et 1 (artefacts possibles hors plage)
    return np.clip(img_k, 0, 1)

# ── Calcul du ratio de compression pour chaque rang
def compression_ratio(m, n, k):
    """Proportion d'éléments stockés par rapport à l'image originale."""
    return k * (m + n + 1) / (m * n)

ranks = [10, 50, 100, len(S_img)]   # 4 rangs à tester
for k in ranks:
    ratio = compression_ratio(m, n, k)
    img_k = svd_compress(U_img, S_img, Vt_img, k)
    err   = np.linalg.norm(img_gray - img_k, 'fro')
    psnr  = 20 * np.log10(1.0 / np.sqrt(np.mean((img_gray-img_k)**2)))
    print(f"rang={k:4d} | stockage={ratio*100:5.1f}% | ‖err‖_F={err:7.3f} | PSNR={psnr:6.2f} dB")
```

4.1

Précision : les méthodes directes (LU, Cholesky) atteignent systématiquement la précision machine (résidu ≈ 2–4 × 10⁻¹⁶ ≈ ε_mach). Elles résolvent le système de façon algébriquement exacte à l'arrondi flottant près. Les méthodes itératives s'arrêtent dès que la correction ‖x⁽ᵏ⁺¹⁾ − x⁽ᵏ⁾‖ < ε = 10⁻⁶, ce qui donne un résidu légèrement supérieur (~10⁻⁶ pour Jacobi, ~10⁻⁷ pour Gauss-Seidel).

Temps : pour notre matrice 4×4 dense, LU ≈ 67 µs, Cholesky ≈ 28 µs, et les méthodes itératives exécutent 10–19 itérations à O(n²) chacune. Pour des matrices creuses de grande taille (n ~ 10⁴–10⁶), les méthodes itératives deviennent indispensables : chaque itération ne coûte que O(nnz) opérations (nnz = nombre d'éléments non nuls), alors que la factorisation LU dense coûte O(n³) et peut nécessiter O(n²) de mémoire.

Conclusion : matrices denses petites/moyennes → LU ou Cholesky (précision machine garantie). Matrices creuses large-échelle ou systèmes d'EDP discrétisées → Jacobi, Gauss-Seidel, GMRES, ou gradient conjugué.

4.2


Expliquez l'impact de la réduction de dimension via SVD sur l'analyse de données.
Théorème d'Eckart-Young : l'approximation rang-k est optimale. Parmi toutes les matrices de rang ≤ k, D_k = U_k Σ_k V_kᵀ minimise l'erreur au sens des normes de Frobenius et spectrale. SVD garantit la meilleure compression possible à rang donné.

Impact concret sur D : σ₁ = 92.87 capte 72.7% de l'énergie — les notes des 10 étudiants sont fortement corrélées entre matières. La «force générale» d'un étudiant explique la majorité des variations. Garder 2 composantes (83.6% d'énergie) suffit pour une visualisation et une analyse exploratoire.

Effets positifs : réduction du bruit (les petites σ capturent souvent du bruit), suppression de la multicolinéarité, réduction de la malédiction de la dimension, découverte de structures latentes. Limite : une erreur de Frobenius de 14.1 (rang-2) sur D signifie que certaines nuances individuelles sont perdues. Le choix du rang est un compromis fidélité/complexité.



4.3


Discutez des avantages et limites de l'ACP pour l'analyse de données.
Avantages : l'ACP est non supervisée, sans hyperparamètre critique, et produit des composantes orthogonales (décorrélées) qui éliminent la multicolinéarité. Elle maximise la variance préservée composante par composante (propriété d'optimalité de la SVD). Elle réduit le coût des algorithmes aval (k-means, SVM, régression) et offre une visualisation naturelle 2D/3D. Pour notre matrice D, les 2 premières CP capturent 63.5% de la variance — synthèse exploitable.

Limites fondamentales : l'ACP est linéaire — elle ne capte pas les structures non linéaires (spirales, clusters en cercles concentriques, manifolds courbes). Elle est sensible à l'échelle : sans standardisation préalable, les matières à grande variance dominent artificiellement. Les composantes principales ne sont pas nécessairement interprétables : CP1 est une combinaison linéaire de toutes les matières, pas une matière spécifique. L'ACP suppose implicitement que la variance est un proxy de l'information utile — or pour certaines tâches (classification rare), les directions de faible variance peuvent être discriminantes.

Alternatives pour dépasser ces limites : t-SNE et UMAP pour les structures non linéaires (mais sans reconstruction), Kernel PCA pour appliquer l'ACP dans un espace de features implicite via un noyau RBF ou polynomial, ICA pour extraire des sources indépendantes (non gaussiennes) plutôt que décorrélées.
